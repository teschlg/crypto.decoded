## Encoding

If we want to encrypt text, the appropriate data type is the string `str`. In cryptography, however, we want to encrypt more than just text, so we assume that the plaintext (that is, the data to be encrypted) is given as a sequence of bytes (an octet string). The appropriate data type in Python is the byte string `bytes`. It is entered in the form

```python
x = b"abc"
type(x)
```

The conversion is performed by mapping each character to its corresponding ASCII code. However, since ASCII uses only 7 bits (that is, values from 0 to 127), not all characters can be represented this way (for example, characters with umlauts cannot be represented). Originally, the remaining 128 possibilities were therefore used for commonly occurring special characters. However, this is
not sufficient to cover all special characters, so different encodings (mappings) were used for different languages:
For example, ISO 8859 Latin-1 for Western European languages, Latin-2 for Central European languages, etc.

The conversion is performed using `.encode(...)`

```python
x = "Österreich"
y = x.encode("latin1")
y
```

These encodings are compatible with ASCII, and Python represents all characters that cannot be displayed as ASCII characters as hexadecimal numbers. Thus, \xd6 = 0xD6 = 214.

These characters can only be displayed once the encoding is known, using:

```python
y.decode("latin1")
```

If we want a single encoding to cover all special characters, we need more than one byte per character.
The current standard is Unicode UTF-8 (or UTF-16, UTF-32), and Python assumes UTF-8 when no encoding is specified.

```python
x.encode()
```

As you can see, a character in the original string may now correspond to more than one byte.

Summary:
* An encoding is required to convert strings into byte strings
* Different encodings produce different byte strings
* With UTF-8, the encoding commonly used today, a character may correspond to multiple bytes

Finally, a few comments on working with byte strings:

When we iterate over a byte string, we obtain the individual bytes as numbers:

```python
x = b"abc"
for i in x:
    print(i)
```

Caution: A byte string cannot be modified (it is immutable). If we want to modify a byte string, we must use a `bytearray`:

```python
x = bytearray(x)
x[0] = 65
x
```

```python
x.append(66)
x
```

It can be converted back using

```python
bytes(x)
```

For an individual character, we can use

```python
ord("a")
```

and

```python
chr(_)
```

.

To transmit byte strings over channels that support only ASCII (email, paper, the Web, etc.), it is common to transmit short byte strings as hexadecimal numbers:

```python
y.hex()
```

The reverse conversion is performed using

```python
bytes.fromhex(_)
```

Base64 is used for longer byte strings

```python
from base64 import b64decode, b64encode

b64encode(y)
```

Especially with longer text, it is convenient that it can be split across lines without interfering with decoding:

```python
b64decode(
    """1nN0ZXJyZ
WljaA=="""
)
```
