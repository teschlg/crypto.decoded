## Codierung

Wollen wir einen Text verschlüsseln, so ist der passende Datentyp der String `str`. In der Kryptographie möchte man aber nicht nur Texte verschlüsseln und geht daher davon aus, dass der Klartext (also die zu verschlüsselnden Daten) als eine Folge von Bytes (Octet-String) gegeben ist. Der passende Datentyp in Python ist der Bytestring `bytes`. Die Eingabe erfolgt in der Form

```python
x = b"abc"
type(x)
```

Die Umwandlung erfolgt, indem jedes Zeichen seinem zugehörigen ASCII-Code zugeordnet wird. Da ASCII nur 7 Bit (also 0 bis 127) verwendet, können damit aber nicht alle Zeichen dargestellt werden (z. B. keine Umlaute). Ursprünglich wurden deshalb die restlichen 128 Möglichkeiten für gängige Sonderzeichen verwendet. Das reicht aber
nicht aus, um alle Sonderzeichen abzudecken, und man hat für verschiedene Sprachen verschiedene Codierungen (Zuordnungen) verwendet:
z. B. ISO 8859 Latin-1 für Westeuropäisch, Latin-2 für Mitteleuropäisch etc.

Die Umwandlung erfolgt mit `.encode(...)`

```python
x = "Österreich"
y = x.encode("latin1")
y
```

Diese Codierungen sind kompatibel zu ASCII und Python stellt dabei alle Zeichen, die nicht als ASCII-Zeichen darstellbar sind, als Hexzahlen dar. Also \xd6 = 0xD6 = 214.

Eine Darstellung dieser Zeichen ist erst möglich, wenn die Codierung bekannt ist, und erfolgt mit:

```python
y.decode("latin1")
```

Will man alle Sonderzeichen mit einer Codierung abdecken, so benötigt man mehr als ein Byte pro Zeichen.
Der aktuelle Standard ist Unicode UTF-8 (bzw. UTF-16, UTF-32), wobei Python von UTF-8 ausgeht, wenn keine Codierung angegeben wird.

```python
x.encode()
```

Wie Sie sehen, entspricht ein Zeichen im ursprünglichen String jetzt unter Umständen mehr als einem Byte.

Zusammenfassung:
* Um Strings in Bytestrings umzuwandeln, ist eine Codierung notwendig.
* Verschiedene Codierungen ergeben verschiedene Bytestrings.
* Bei der aktuell gängigen Codierung UTF-8 entspricht ein Zeichen eventuell mehreren Bytes.

Zum Abschluss noch einige Bemerkungen zum Umgang mit Bytestrings:

Wenn wir über einen Bytestring iterieren, erhalten wir die einzelnen Bytes als Zahlen:

```python
x = b"abc"
for i in x:
    print(i)
```

Achtung: Ein Bytestring ist nicht veränderbar (immutable). Wenn wir einen Bytestring ändern wollen, müssen wir ein `bytearray` verwenden:

```python
x = bytearray(x)
x[0] = 65
x
```

```python
x.append(66)
x
```

Rückumwandlung erfolgt mit

```python
bytes(x)
```

Für ein einzelnes Zeichen können wir

```python
ord("a")
```

und

```python
chr(_)
```

verwenden.

Um Bytestrings über Kanäle zu übertragen, die nur ASCII beherrschen (E-Mail, Papier, Web etc.), ist es oft üblich, bei kurzen Strings den Bytestring als Hexzahl zu übertragen:

```python
y.hex()
```

Retour geht es mit

```python
bytes.fromhex(_)
```

Für längere Bytestrings verwendet man Base64.

```python
from base64 import b64decode, b64encode

b64encode(y)
```

Gerade bei längeren Texten ist es praktisch, dass diese auch aufgeteilt werden dürfen, ohne dass die Decodierung darunter leidet:

```python
b64decode(
    """1nN0ZXJyZ
WljaA=="""
)
```
